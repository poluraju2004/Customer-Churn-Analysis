```python
plt.figure(figsize=(6, 4))

df["Churn"].value_counts().sort_index().plot(kind="bar")

plt.title("Customer Churn Distribution")
plt.xlabel("Churn")
plt.ylabel("Number of Customers")
plt.savefig("C:/Users/polur/Desktop/Customer Churn Analysis/Charts/Bar_Churn.png")

plt.show()
```

```python
plt.figure(figsize=(6, 4))

sns.countplot(
    data=df,
    x="Churn"
)

plt.title("Customer Churn Distribution")
plt.xlabel("Churn")
plt.ylabel("Number of Customers")
plt.savefig("C:/Users/polur/Desktop/Customer Churn Analysis/Charts/CountPlot_Churn.png")
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.boxplot(
    data=df,
    x="Churn",
    y="tenure"
)

plt.title("Tenure by Churn")
plt.xlabel("Churn")
plt.ylabel("Tenure (Months)")
plt.savefig("C:/Users/polur/Desktop/Customer Churn Analysis/Charts/BoxPlot_Churn._tenure.png")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.boxplot(
    data=df,
    x="Churn",
    y="MonthlyCharges"
)

plt.title("Monthly Charges by Churn")
plt.xlabel("Churn")
plt.ylabel("Monthly Charges")
plt.savefig("C:/Users/polur/Desktop/Customer Churn Analysis/Charts/BoxPlot_Churn_MonthlyCharges.png")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x="tenure",
    y="MonthlyCharges",
    hue="Churn"
)

plt.title("Tenure vs Monthly Charges")
plt.xlabel("Tenure (Months)")
plt.ylabel("Monthly Charges")
plt.savefig("C:/Users/polur/Desktop/Customer Churn Analysis/Charts/BoxPlot_tenure_MonthlyCharges.png")

plt.show()
```

```python
churn_counts = df["Churn"].value_counts()

plt.figure(figsize=(7, 7))

plt.pie(
    churn_counts,
    labels=["Not Churned", "Churned"],
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Customer Churn Percentage")
plt.savefig("C:/Users/polur/Desktop/Customer Churn Analysis/Charts/Piechart_Churn.png")


plt.show()
```

```python
numeric_columns = [
    "SeniorCitizen",
    "tenure",
    "MonthlyCharges",
    "Churn"
]

correlation = df[numeric_columns].corr()

plt.figure(figsize=(8,6))

sns.heatmap(
    correlation,
    annot=True,
    fmt=".2f"
)

plt.title("Correlation Heatmap")
plt.savefig("C:/Users/polur/Desktop/Customer Churn Analysis/Charts/correlation_heatmap.png",
            dpi=300,
            bbox_inches="tight")

plt.show()
```

```python
tenure_churn = df.groupby("tenure")["Churn"].mean() * 100

plt.figure(figsize=(10,5))

plt.plot(
    tenure_churn.index,
    tenure_churn.values
)

plt.title("Churn Rate by Tenure")
plt.xlabel("Tenure (Months)")
plt.ylabel("Churn Rate (%)")
plt.savefig("C:/Users/polur/Desktop/Customer Churn Analysis/Charts/plot_Churn.png")

plt.show()
```

```python
plt.figure(figsize=(7,5))

sns.countplot(
    data=df,
    x="gender",
    hue="Churn"
)

plt.title("Churn by Gender")
plt.xlabel("Gender")
plt.ylabel("Number of Customers")

plt.tight_layout()
plt.savefig("C:/Users/polur/Desktop/Customer Churn Analysis/Charts/CountPlot_gender",dpi=300)
plt.show()
```
